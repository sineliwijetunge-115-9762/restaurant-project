### Week 1 work

##### The datasets are imported into the notebook using pandas. Altogether 6 datasets are used. 

```python
D1 = pd.read_csv("yelp_reviews.csv")
D1.dropna(how='all')
D1.drop_duplicates()
```

```python
D2 = pd.read_json("yelp_academic_dataset_business.json", lines=True)
D2.dropna(how='all')
```

```python
D3 = pd.read_json("yelp_academic_dataset_checkin.json", lines=True)
D3.dropna(how='all')
```

```python
D4 = pd.read_json("yelp_academic_dataset_review.json", lines=True, nrows=140000)
D4.dropna(how='all')
```

```python
D5 = pd.read_json("yelp_academic_dataset_tip.json", lines=True, nrows=140000)
D5.dropna(how='all')
```

```python
D6 = pd.read_json("yelp_academic_dataset_user.json", lines=True, nrows=140000)
D6.dropna(how='all')
```

##### The datasets we collected contains details of many businesses, not only food sector. Therefore, a new dataframe 'Res' is created to store details of only restaurants/cafes/bars..(Basically any place that sell food). The catergorisation is done with the help of the 'categories' field in the 'D2' dataframe. The below piece of code checks if the word 'food' is found in 'catergories' column, if found the record is added to the dataframe 'Res'. I have also made sure that null values will be dropped. (dropna used)

```python
Res = D2[D2['categories'].str.contains('Food', case=False, na=False)]
Res = Res[Res['is_open']==1]
Res.dropna(how='all')
```

##### A new dataframe named 'Combined_Res' is created to merge all the datasets together. As the 'business_id' is common in all the datasets, that is used to merge.  

```python
Combined_Res = pd.merge(Res, D1, on = 'business_id', how='left')
Combined_Res.dropna(how='all')
```

```python
Combined_Res = pd.merge(Res, D4[['text', 'business_id']], on='business_id', how='left')
Combined_Res.dropna()
```

```python
Combined_Res = pd.merge(Res, D5[['text', 'business_id']], on='business_id', how='left')
Combined_Res.dropna()
```

##### Below code counts how many reviews each restaurant/cafe/bar has in total according to the data collected. 'groupby' is used along with sum to complete this task. I have also made sure that the reviews are in descending order. So, the place with highest reviews will appear first. I also included the average ratings by taking the average of the number of stars using the mean function so that even though if there are less reviews a decision can be made by looking at the average rating. aggregate function (.agg) is used to make it possible to perform two different mathematical operations to two different columns inside groupby function. 

```python
Reviews = (Combined_Res.groupby('name', as_index=False)
           .agg(
               Total_Reviews=('review_count', 'sum'),
               Avg_Rating=('stars', 'mean')
           )
           .sort_values(by='Total_Reviews', ascending=False))
print(Reviews)
```

### Week 2 work

##### Below code helps cluster places by its geographical locations. I have used sklearn.cluster to import KMeans. Then I made use of the latitude and longitude columns in the dataset as coordinates. Then based on those coordinates a scatter plot was plotted. I have used n = 5 clusters. The density of clusters suggest the number of businesses in that particular area. Less density (small circle) indicates less businesses in the area and vise-versa.

```python
from sklearn.cluster import KMeans

coords = Combined_Res[['latitude', 'longitude']]

kmeans = KMeans(n_clusters = 5)
Combined_Res['location_cluster'] = kmeans.fit_predict(coords)

plt.scatter(Combined_Res['longitude'],Combined_Res['latitude'],c=Combined_Res['location_cluster'])
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Location clusters")
plt.show()
```
